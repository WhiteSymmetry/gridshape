generate_shape: kare, dikdörtgen, daire, üçgen, elmas, çokgen

```python
from gridshape import GridShape

# Interactive munu: Interaktif menüyü başlatmak için:
GridShape.run_interactive()
```

```python
from gridshape import GridShape
import matplotlib.pyplot as plt

g = GridShape(outer_size=10)
g.generate_shape('square', {'row': 5, 'col': 5, 'h': 3, 'w': 3}) # Kare
g.visualize()
plt.show()
```

```python
from gridshape import GridShape
import matplotlib.pyplot as plt

g = GridShape(outer_size=10)
g.generate_shape('triangle', {'row': 5, 'col': 5, 'size': 3}) # Üçgen
g.visualize()
plt.show()
```

```python
from gridshape import GridShape
import matplotlib.pyplot as plt

g = GridShape(outer_size=10)
g.generate_shape('circle', {'row': 5, 'col': 5, 'radius': 3}) # Daire
g.visualize()
plt.show()
```

```python
from gridshape import GridShape
import matplotlib.pyplot as plt

g = GridShape(outer_size=10)
g.generate_shape('diamond', {'row': 5, 'col': 5, 'size': 3}) # Elmas
g.visualize()
plt.show()
```

```python
from gridshape import GridShape
import matplotlib.pyplot as plt

g = GridShape(outer_size=10)
g.generate_shape('polygon', {'coords': [(2,2), (5,2), (5,5)]})  # Üçgen çokgen
g.visualize()
plt.show()
```
